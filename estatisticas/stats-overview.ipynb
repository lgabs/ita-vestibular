# Estatísticas do Vestibular do ITA
Este notebook carrega os dados extraídos do site do ITA e gera alguns gráficos históricos.


```
import json
import pandas as pd
import matplotlib.pyplot as plt
plt.style.use('fivethirtyeight')

with open('estatisticas/estatisticas.json', 'r', encoding='utf-8') as fh:
    data = json.load(fh)
```

```
# Extrair quantidade de inscritos por categoria
records = []
for year, tables in data.items():
    for item in tables:
        if item['heading'] and 'Número de candidatos inscritos' in item['heading']:
            df = pd.DataFrame(item['table'][1:], columns=item['table'][0])
            df['year'] = int(year)
            records.append(df)
inscritos = pd.concat(records)
inscritos.head()
```

```
# Transformar colunas numericas
for col in ['Homens','Mulheres','Total']:
    inscritos[col] = inscritos[col].str.replace('.', '', regex=False)
    inscritos[col] = inscritos[col].str.replace(',', '.', regex=False)
    inscritos[col] = pd.to_numeric(inscritos[col], errors='coerce')
inscritos
```

```
# Plotar total de inscritos por tipo
pivot = inscritos.pivot_table(index='year', columns='Tipo', values='Total', aggfunc='first')
pivot.sort_index(inplace=True)
ax = pivot.plot(marker='o', figsize=(10,6))
ax.set_ylabel('Número de Inscritos')
ax.set_xlabel('Ano')
ax.set_title('Candidatos inscritos por ano')
plt.tight_layout()
```
